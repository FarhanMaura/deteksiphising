```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

print("Menyiapkan data untuk Genetic Algorithm...")
# Kita pakai sampel 10.000 baris agar komputasi GA tidak memakan waktu berjam-jam
X_sample = X.sample(n=10000, random_state=42)
y_sample = y.loc[X_sample.index]

X_train_ga, X_test_ga, y_train_ga, y_test_ga = train_test_split(X_sample, y_sample, test_size=0.2, random_state=42)

# Konfigurasi Genetic Algorithm Sederhana
jumlah_populasi = 10
jumlah_generasi = 3 # Dibuat 3 generasi agar cepat selesai
jumlah_fitur = X.shape[1] # 50 fitur asli

# 1. Inisialisasi Populasi Awal (Kombinasi biner 0 dan 1 secara acak)
populasi = np.random.randint(2, size=(jumlah_populasi, jumlah_fitur))

def hitung_fitness(kromosom):
    fitur_terpilih = np.where(kromosom == 1)[0]
    if len(fitur_terpilih) == 0: return 0.0

    # Evaluasi menggunakan Random Forest berskala kecil
    model = RandomForestClassifier(n_estimators=10, random_state=42)
    model.fit(X_train_ga.iloc[:, fitur_terpilih], y_train_ga)
    prediksi = model.predict(X_test_ga.iloc[:, fitur_terpilih])
    return accuracy_score(y_test_ga, prediksi)

print("Mulai proses evolusi Genetic Algorithm...\n")
best_kromosom_global = None
best_fitness_global = 0

# 2. Proses Evolusi (Seleksi)
for generasi in range(jumlah_generasi):
    nilai_fitness = [hitung_fitness(ind) for ind in populasi]

    best_idx = np.argmax(nilai_fitness)
    if nilai_fitness[best_idx] > best_fitness_global:
        best_fitness_global = nilai_fitness[best_idx]
        best_kromosom_global = populasi[best_idx]

    print(f"Generasi {generasi + 1} | Akurasi Terbaik: {best_fitness_global * 100:.2f}%")

    # (Simulasi Crossover dan Mutasi Sederhana untuk generasi berikutnya)
    populasi_baru = []
    for _ in range(jumlah_populasi):
        parent1, parent2 = populasi[np.random.choice(jumlah_populasi, 2, replace=False)]
        titik_potong = np.random.randint(1, jumlah_fitur-1)
        anak = np.concatenate([parent1[:titik_potong], parent2[titik_potong:]])
        # Mutasi acak
        if np.random.rand() < 0.1:
            titik_mutasi = np.random.randint(jumlah_fitur)
            anak[titik_mutasi] = 1 - anak[titik_mutasi]
        populasi_baru.append(anak)
    populasi = np.array(populasi_baru)

# 3. Hasil Akhir Seleksi Fitur
fitur_terpilih_idx = np.where(best_kromosom_global == 1)[0]
nama_fitur_terpilih = X.columns[fitur_terpilih_idx].tolist()

print("\n==================================================")
print("PROSES GENETIC ALGORITHM SELESAI!")
print("==================================================")
print(f"Jumlah fitur awal: {jumlah_fitur}")
print(f"Jumlah fitur yang TERPILIH: {len(nama_fitur_terpilih)}")
print("\nDaftar Fitur yang Terpilih untuk dimasukkan ke Tabel 3.3:")
for i, fitur in enumerate(nama_fitur_terpilih, 1):
    print(f"{i}. {fitur}")

# Buat dataset baru yang HANYA berisi fitur terpilih
X_optimized = X[nama_fitur_terpilih]
```

```python
# 1. Buang kolom yang isinya cuma teks/identitas (nggak bisa diolah Machine Learning)
kolom_teks = ['FILENAME', 'URL', 'Domain', 'TLD', 'Title']
df_clean = df.drop(columns=kolom_teks, errors='ignore')

# 2. Hapus baris yang ada missing value-nya (NaN)
df_clean = df_clean.dropna()

# 3. Misahin Fitur (X) dan Target/Label (y)
# Kita pakai .iloc (lokasi indeks) buat ngindarin error nama kolom 'label' yang berantakan
X = df_clean.iloc[:, :-1]  # Ambil semua kolom KECUALI kolom paling akhir
y = df_clean.iloc[:, -1]   # Ambil KHUSUS kolom paling akhir aja (label)

# 4. Pastiin datanya udah bentuk angka numerik semua buat jaga-jaga
X = X.apply(pd.to_numeric, errors='coerce')
X = X.dropna(axis=1, how='any') # Buang kolom yang gagal diconvert ke angka

print("Data berhasil dibersihkan dan dipisah!")
print(f"Total dimensi data Fitur (X) sekarang: {X.shape}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Drop fitur yang memiliki korelasi ekstrim (Target Leakage) untuk mencegah Overfitting
cols_to_drop = ['URLSimilarityIndex', 'DomainTitleMatchScore', 'URLTitleMatchScore']
X_final = X_optimized.drop(columns=cols_to_drop, errors='ignore')

print(f"Sisa fitur optimal yang digunakan: {X_final.shape[1]}")
print("Membagi data latih dan data uji (80:20)...")

# Membagi data menggunakan X_final yang sudah difilter
X_train_rf, X_test_rf, y_train_rf, y_test_rf = train_test_split(X_final, y, test_size=0.2, random_state=42)

print(f"Jumlah Data Latih: {X_train_rf.shape[0]} baris")
print(f"Jumlah Data Uji: {X_test_rf.shape[0]} baris\n")

# Inisialisasi Model Random Forest
rf_model = RandomForestClassifier(n_estimators=100, max_depth=None, criterion='gini', random_state=42)

# ======================================================
# SIAP-SIAP SCREENSHOT BAGIAN OUTPUT INI YA BUAT BAB 4
# ======================================================
print("======================================================")
print("Memulai proses pelatihan (Training) model Random Forest...")
print("======================================================")
rf_model.fit(X_train_rf, y_train_rf)
print(">> Status: Pelatihan Model Selesai!\n")

# Cek hasil akurasi tesnya
y_pred_rf = rf_model.predict(X_test_rf)
akurasi_rf = accuracy_score(y_test_rf, y_pred_rf)
print(f"Akurasi Model Prediksi (Random Forest): {akurasi_rf * 100:.2f}%")
```
